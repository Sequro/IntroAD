```python
#Proyecto graf
import pandas as pd
import matplotlib.pyplot as plt

datos = pd.read_csv("climate-change_cri.csv")

df = datos[["Country Name",
            "Country ISO3",
            "Year",
            "Indicator Name",
            "Indicator Code", "Value"]]
df = df.rename(columns={
    "Country Name": "PAIS",
    "Country ISO3": "ISO3",
    "Year": "AÑO",
    "Indicator Name": "INDICADOR",
    "Indicator Code": "CODIGO_INDICADOR",
    "Value": "VALOR"
})
areas = ['C02 emissions from gaseous fuel consumption (kt)',
         'CO2 emissions from liquid fuel consumption (kt)',
         'CO2 emissions from solid fuel consumption (kt)',
         'Methane emissions (kt of CO2 equivalent)',
         'Nitrous oxide emissions (kt of CO2 equivalent)',
         'Total greenhouse gas emissions (kt of CO2 equivalent)'
         ]
df['VALOR'] = pd.to_numeric(df['VALOR'], errors='coerce')
df = df[df['INDICADOR'].isin(areas)]
df_grouped = df.groupby(['AÑO', 'INDICADOR'])['VALOR'].sum().unstack()
df_grouped.plot.bar(figsize=(15, 7))
plt.xlabel('Año')
plt.ylabel('Valor (kt o kt de CO2 equivalente)')
plt.title('Emisiones de Gases de Efecto Invernadero por Tipo')
plt.legend(title='Indicador')
plt.show()
```
