```python
from geopy.geocoders import Nominatim
geolocator = Nominatim(user_agent = "stemcoder")

Location = geolocator.geocode("Costa Rica")

if Location is not None:
  print("Ubicación encontrada:")
  print("Dirección: ", Location.address)
  print("Latitud: ", Location.latitude)
  print("Longitud: ", Location.longitude)
else:
  print("Ubicación no encontrada")
```

```python
from google.colab import files
import folium

lat, lon = 9.7489, -83.7534

costa_rica_map = folium.Map(location=[lat,lon], zoom_start=7)
costa_rica_map.save("CostaRica.html")
files.download("CostaRica.html")

costa_rica_map
```

```python
campuses = [ "Campus Omar Dengo","Campus Benjamín Núñez", "Sede Interuniversitaria", "Campus Sarapiquí", "Campus Coto",
            "Campus Pérez Zeledón", "Campus Liberia", "Campus Nicoya"]

for x in campuses:
  locForE = geolocator.geocode(x)
  if location is not None:
    print("Ubicación encontrada:")
    print("Dirección: ", locForE.address)
    print("Latitud: ", locForE.latitude)
    print("Longitud: ", locForE.longitude, "\n")
  else:
    print("Ubicación no encontrada")
```

```python
import folium
from folium.plugins import MiniMap
from branca.element import Figure


my_map = folium.Map(location=(9.7489, -83.7534))

for x in campuses:

  locForEMarks = geolocator.geocode(x)

  name = locForEMarks.address
  popuptext = "<b>"+name+"</>"
  folium.Marker(location=[locForEMarks.latitude, locForEMarks.longitude], popup= popuptext, icon=folium.Icon(color='blue',icon="ok.sing")).add_to(my_map)

  folium.Circle(location=[locForEMarks.latitude, locForEMarks.longitude], color='red', weight=6,fill_color='purple', fill_opacity=0.5, tooltip=name).add_to(my_map)

my_map
```
